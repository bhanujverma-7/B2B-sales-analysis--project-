# Cleaning 
(Null/Duplicate check, remove unneccessary columns, rename cols.)

```python
new_header = state_reg.iloc[0]
state_reg.columns = new_header
state_reg = state_reg[1:]

state_reg.head()
```

Null check

```python
print('Missing values in sales_orders:\n', sales_orders.isnull().sum(), '\n')
print('Missing values in customers:\n', customers.isnull().sum(), '\n')
print('Missing values in regions:\n', regions.isnull().sum(), '\n')
print('Missing values in products:\n', products.isnull().sum(), '\n')
print('Missing values in state_reg:\n', state_reg.isnull().sum(), '\n')
#print('Missing values in budgets:\n', budgets.isnull().sum(), '\n')
```

```python
df = sales_orders.merge(customers,how = 'left',left_on = 'Customer Name Index', right_on='Customer Index')
df = df.merge(products,how = 'left',left_on = 'Product Description Index', right_on='Index')
df = df.merge(regions,how = 'left',left_on = 'Delivery Region Index', right_on='id')
df = df.merge(state_reg[['State Code', 'Region']],how = 'left',left_on = 'state_code', right_on='State Code')
# df = df.merge(budgets,how = 'left',left_on = 'Delivery Region Index', right_on='Region')
```

```python
df.drop(columns=['latitude', 'longitude', 'area_code','land_area', 'water_area', 'time_zone'], axis=1, inplace=True)
```

# Featue Engineering

```python
# Important Features
df['total_mfg_cost'] = df['mfg_cost(USD)'] * df['order_quantity']
df['profit'] = df['revenue(USD)'] - df['total_mfg_cost']
df['profit_margin(%)'] = (df['profit'] / df['revenue(USD)'])*100
```

**Column to show gap between purchases**

```python
df = df.sort_values(['customer_name', 'orderdate'])
```

```python
df['Previous_Purchase_Date'] = (
    df.groupby('customer_name')['orderdate']
      .shift(1)
)
```

```python
df['Gap_Days'] = (
    df['orderdate'] - df['Previous_Purchase_Date']
).dt.days
```

```python
df.head()
```

```python
customer_summary = (
    df.groupby('customer_name')
      .agg(
          Total_Orders=('orderdate', 'count'),
          Avg_Gap_Days=('Gap_Days', 'mean'),
          Min_Gap=('Gap_Days', 'min'),
          Max_Gap=('Gap_Days', 'max')
      )
      .reset_index()
)
```
